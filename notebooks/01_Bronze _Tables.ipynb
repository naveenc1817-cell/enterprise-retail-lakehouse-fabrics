```python
from pyspark.sql.functions import current_timestamp, input_file_name

base_path = "Files/bronze"

files = {"customers": "olist_customers_dataset.txt",
    "orders": "olist_orders_dataset.txt",
    "order_items": "olist_order_items_dataset.txt",
    "payments": "olist_order_payments_dataset.txt",
    "products": "olist_products_dataset.txt",
    "sellers": "olist_sellers_dataset_Clean.txt",
    "geolocation": "olist_geolocation_dataset.txt",
    "category_translation": "product_category_name_translation.txt"}

for table_name, file_name in files.items():
    df = (spark.read
        .option("header", "true")
        .option("inferSchema", "true")
        .option("delimiter", ",")
        .csv(f"{base_path}/{file_name}"))

    df = (df.withColumn("bronze_load_timestamp", current_timestamp())
          .withColumn("source_file_name", input_file_name()))

    df.write.mode("overwrite").format("delta").saveAsTable(f"bronze_{table_name}")

    print(f"Created bronze_{table_name}: {df.count()} rows")
```
